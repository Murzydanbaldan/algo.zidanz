```python
# Import Library
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset penguins dari seaborn
penguins = sns.load_dataset("penguins")

# Boxplot Tunggal
sns.boxplot(x=penguins['bill_length_mm'])
plt.show()

sns.boxplot(y=penguins['bill_length_mm'])
plt.show()

# Boxplot dengan Beberapa Kategori
sns.boxplot(x=penguins["species"], y=penguins['bill_length_mm'])
plt.show()

# Pengaturan Warna
sns.boxplot(data=penguins, x="species", y='bill_length_mm', color="orange")
plt.show()

# Untuk mengatur warna palette
sns.boxplot(data=penguins, x="species", y='bill_length_mm', palette="bright")
plt.show()

# Pengaturan warna dengan custom palette
custom_pal = {"Adelie": "#9A0EEA", "Chinstrap": "beige", "Gentoo": "c"}
sns.boxplot(data=penguins, x="species", y='bill_length_mm', palette=custom_pal)
plt.show()

# Pengaturan Ukuran dan Bentuk Boxplot
sns.boxplot(
    data=penguins, x="species", y='bill_length_mm',
    palette=custom_pal,
    linewidth=0.7,
    width=0.4
)
plt.show()

# Dengan notch, ukuran outlier lebih besar
sns.boxplot(
    data=penguins, x="species", y='bill_length_mm',
    palette=custom_pal,
    linewidth=0.7,
    width=0.4,
    notch=True,
    fliersize=10,
    flierprops={"marker": "o"}
)
plt.show()

# Boxplot dengan Subkategori
plt.figure(figsize=(8, 6))
sns.boxplot(
    data=penguins, x="species", y='bill_length_mm', hue="sex",
    palette="bright",
    linewidth=0.8,
    width=0.5,
    fliersize=3,
    flierprops={"marker": "o"}
)
plt.legend(bbox_to_anchor=(1, 0.6), loc=2, borderaxespad=1)
plt.show()

# Menampilkan Beberapa Diagram dalam Satu Frame (1x2)
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(8, 4), sharey=False)
sns.boxplot(data=penguins, x="species", y='bill_length_mm', palette="bright", width=0.6, ax=axes[0])
sns.boxplot(data=penguins, x="species", y='bill_depth_mm', palette="bright", width=0.6, ax=axes[1])
fig.tight_layout()
plt.show()

# Menampilkan Beberapa Diagram dalam Satu Frame (2x2)
fig, axes = plt.subplots(2, 2, figsize=(8, 6), sharex=False)

sns.boxplot(data=penguins, x="species", y='bill_length_mm', palette="bright", width=0.6, ax=axes[0, 0])
sns.boxplot(data=penguins, x="species", y='bill_depth_mm', palette="bright", width=0.6, ax=axes[0, 1])
sns.boxplot(data=penguins, x="species", y='flipper_length_mm', palette="bright", width=0.6, ax=axes[1, 0])
sns.boxplot(data=penguins, x="species", y='body_mass_g', palette="bright", width=0.6, ax=axes[1, 1])

fig.tight_layout()
plt.show()

# Menampilkan Beberapa Diagram dengan Judul
fig, axes = plt.subplots(2, 2, figsize=(8, 6), sharex=False)

p1 = sns.boxplot(data=penguins, x="species", y='bill_length_mm', palette="bright", width=0.6, ax=axes[0, 0])
p2 = sns.boxplot(data=penguins, x="species", y='bill_depth_mm', palette="bright", width=0.6, ax=axes[0, 1])
p3 = sns.boxplot(data=penguins, x="species", y='flipper_length_mm', palette="bright", width=0.6, ax=axes[1, 0])
p4 = sns.boxplot(data=penguins, x="species", y='body_mass_g', palette="bright", width=0.6, ax=axes[1, 1])

plot = [p1, p2, p3, p4]
label = ["Bill Length (mm)", "Bill Depth (mm)", "Flipper Length (mm)", "Body Mass (g)"]

# Set judul untuk masing-masing subplot
for i in range(0, 4):
    plot[i].set(ylabel=None, xlabel=None)
    plot[i].axes.set_title(label[i], fontsize=11)

fig.tight_layout()
plt.show()
```
